```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import GroupShuffleSplit

df = pd.read_csv("cleaned.csv")

print("Dataset shape:", df.shape)
print("\nRecords by year:")
print(df["period"].value_counts().sort_index())
```

```python
# Step 02 — Reproduce the EDA split
#
# Purpose:
# Keep 2017 for development and 2018 for final testing.
# Prevent policy overlap between training and validation.

development_df = df.loc[df["period"] == 2017].copy()

# Keep the 2018 target untouched
test_df = df.loc[df["period"] == 2018].drop(
    columns=["target_premium"]
).copy()

splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, val_idx = next(
    splitter.split(
        development_df,
        groups=development_df["ID_policy"]
    )
)

train_df = development_df.iloc[train_idx].copy()
val_df = development_df.iloc[val_idx].copy()

print("Training:", train_df.shape)
print("Validation:", val_df.shape)
print("Test predictors:", test_df.shape)
```

```python
# Step 03 — Split Verification
#
# Purpose:
# Confirm that our new notebook reproduces
# the previously established development split.

train_policies = set(train_df["ID_policy"])
val_policies = set(val_df["ID_policy"])

print("Training records:", len(train_df))
print("Validation records:", len(val_df))
print("Test records:", len(test_df))

print(
    "Shared policies:",
    len(train_policies & val_policies)
)

print(
    "Shared IDs:",
    len(set(train_df["ID"]) & set(val_df["ID"]))
)

assert len(train_df) == 51286
assert len(val_df) == 12169
assert len(test_df) == 59814
assert len(train_policies & val_policies) == 0

print("\nSplit successfully reproduced.")
```

```python
# Phase 2 — Feature Selection
#
# Purpose:
# Apply feature exclusion decisions established during EDA.
# Preserve the original train_df, val_df, and test_df.

target_col = "target_premium"

excluded_columns = [
    # Identifiers
    "ID",
    "ID_policy",
    "ID_insured",

    # Reporting year
    "period",

    # Constant variable
    "lapse",

    # Future-dependent information
    "date_lapse_insured",
    "date_lapse_policy",
    "year_lapse_insured",
    "year_lapse_policy",
    "age_difference",

    # Redundant historical date information
    "date_effect_insured",
    "date_effect_policy",
    "year_effect_insured",
    "year_effect_policy"
]

feature_columns = [
    col for col in train_df.columns
    if col not in excluded_columns + [target_col]
]

print("Original columns:", train_df.shape[1])
print("Excluded predictors:", len(excluded_columns))
print("Selected predictors:", len(feature_columns))
```

```python
# Training predictors and target
X_train = train_df[feature_columns].copy()
y_train = train_df[target_col].copy()

# Validation predictors and target
X_val = val_df[feature_columns].copy()
y_val = val_df[target_col].copy()

# Final test predictors only
X_test = test_df[feature_columns].copy()

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("X_val:", X_val.shape)
print("y_val:", y_val.shape)

print("X_test:", X_test.shape)
```

```python
# Confirm identical feature columns and order
assert X_train.columns.equals(X_val.columns)
assert X_train.columns.equals(X_test.columns)

# Confirm target is excluded
assert target_col not in X_train.columns

# Confirm excluded variables are absent
assert not set(excluded_columns).intersection(X_train.columns)

# Confirm exposure_time is retained
assert "exposure_time" in X_train.columns

print("Feature selection verified successfully.")
```

```python
# Phase 3 — Step 3.1
#
# Purpose:
# Identify missing values in selected training predictors.
# Do not modify X_train, X_val, or X_test.

missing_summary = pd.DataFrame({
    "Missing Count": X_train.isna().sum(),
    "Missing (%)": X_train.isna().mean() * 100,
    "Data Type": X_train.dtypes
})

missing_summary = (
    missing_summary[missing_summary["Missing Count"] > 0]
    .sort_values("Missing (%)", ascending=False)
)

display(missing_summary.round(2))

print("Features with missing values:", len(missing_summary))
print("Total missing cells:", X_train.isna().sum().sum())
```

```python
from sklearn.impute import SimpleImputer

numeric_missing_cols = [
    "C_GI", "C_II",
    "C_IE_P", "C_IE_S", "C_IE_T",
    "C_GE_P", "C_GE_S", "C_GE_T",
    "IICIMUN", "IICIPROV"
]

categorical_missing_cols = ["C_H", "C_C"]
```

```python
# Preserve original feature DataFrames
X_train_imp = X_train.copy()
X_val_imp = X_val.copy()
X_test_imp = X_test.copy()

# Numerical imputation — median
num_imputer = SimpleImputer(strategy="median")

X_train_imp[numeric_missing_cols] = num_imputer.fit_transform(
    X_train[numeric_missing_cols]
)

X_val_imp[numeric_missing_cols] = num_imputer.transform(
    X_val[numeric_missing_cols]
)

X_test_imp[numeric_missing_cols] = num_imputer.transform(
    X_test[numeric_missing_cols]
)

# Categorical imputation — most frequent
cat_imputer = SimpleImputer(strategy="most_frequent")

X_train_imp[categorical_missing_cols] = cat_imputer.fit_transform(
    X_train[categorical_missing_cols]
)

X_val_imp[categorical_missing_cols] = cat_imputer.transform(
    X_val[categorical_missing_cols]
)

X_test_imp[categorical_missing_cols] = cat_imputer.transform(
    X_test[categorical_missing_cols]
)
```

```python
for name, data in [
    ("Training", X_train_imp),
    ("Validation", X_val_imp),
    ("Test", X_test_imp)
]:
    print(f"{name}:")
    print("Shape:", data.shape)
    print("Remaining missing values:", data.isna().sum().sum())
    print()

    assert data.isna().sum().sum() == 0

assert X_train_imp.columns.equals(X_val_imp.columns)
assert X_train_imp.columns.equals(X_test_imp.columns)

print("Simple imputation completed successfully.")
```

```python
imputation_summary = pd.DataFrame({
    "Feature": numeric_missing_cols + categorical_missing_cols,
    "Strategy": (
        ["Median"] * len(numeric_missing_cols)
        + ["Most Frequent"] * len(categorical_missing_cols)
    ),
    "Imputed Value": (
        list(num_imputer.statistics_)
        + list(cat_imputer.statistics_)
    )
})

display(imputation_summary)
```

```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder

# Nominal categorical features
onehot_cols = [
    "type_policy_dg",
    "type_product",
    "distribution_channel",
    "C_C"
]

# Ordered categorical feature
ordinal_cols = ["C_H"]

# Binary categorical features
binary_cols = [
    "gender",
    "type_policy",
    "reimbursement",
    "new_business"
]

# Explicit binary mappings
binary_maps = {
    "gender": {"F": 0, "M": 1},
    "type_policy": {"C": 0, "I": 1},
    "reimbursement": {"No": 0, "Yes": 1},
    "new_business": {"No": 0, "Yes": 1}
}

# Existing numerical features
encoded_source_cols = onehot_cols + ordinal_cols + binary_cols

numeric_cols = [
    col for col in X_train_imp.columns
    if col not in encoded_source_cols
]

# Check feature classification
assert len(encoded_source_cols + numeric_cols) == X_train_imp.shape[1]

print("One-Hot features:", onehot_cols)
print("Ordinal features:", ordinal_cols)
print("Binary features:", binary_cols)
print("Numerical features:", len(numeric_cols))
```

```python
# One-Hot Encoder
onehot_encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False,
    dtype=np.int8
)

onehot_encoder.fit(X_train_imp[onehot_cols])


# Ordinal Encoder for municipality population size
ordinal_encoder = OrdinalEncoder(
    categories=[["H1", "H2", "H3", "H4", "H5", "H6"]],
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

ordinal_encoder.fit(X_train_imp[ordinal_cols])
```

```python
def encode_features(data):

    # 1. Preserve numerical features
    numeric_df = data[numeric_cols].copy()

    # 2. Apply binary encoding
    binary_df = pd.DataFrame(index=data.index)

    for col, mapping in binary_maps.items():
        binary_df[col] = data[col].map(mapping)

        if binary_df[col].isna().any():
            raise ValueError(
                f"Unexpected or missing category found in {col}"
            )

        binary_df[col] = binary_df[col].astype(np.int8)

    # 3. Apply ordinal encoding
    ordinal_values = ordinal_encoder.transform(
        data[ordinal_cols]
    )

    ordinal_df = pd.DataFrame(
        ordinal_values.astype(np.int8) + 1,
        columns=ordinal_cols,
        index=data.index
    )

    # 4. Apply one-hot encoding
    onehot_values = onehot_encoder.transform(
        data[onehot_cols]
    )

    onehot_df = pd.DataFrame(
        onehot_values,
        columns=onehot_encoder.get_feature_names_out(onehot_cols),
        index=data.index
    )

    # 5. Combine all features
    encoded_df = pd.concat(
        [numeric_df, binary_df, ordinal_df, onehot_df],
        axis=1
    )

    return encoded_df


X_train_encoded = encode_features(X_train_imp)
X_val_encoded = encode_features(X_val_imp)
X_test_encoded = encode_features(X_test_imp)

print("Training:", X_train_encoded.shape)
print("Validation:", X_val_encoded.shape)
print("Test:", X_test_encoded.shape)
```

```python
encoding_summary = pd.DataFrame({
    "Feature": X_train_encoded.columns,
    "Data Type": X_train_encoded.dtypes.astype(str)
})

display(encoding_summary)

print("Original predictors:", X_train_imp.shape[1])
print("Encoded predictors:", X_train_encoded.shape[1])
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    OneHotEncoder,
    OrdinalEncoder,
    StandardScaler
)

# Nominal categorical features
onehot_cols = [
    "type_policy_dg",
    "type_product",
    "distribution_channel",
    "C_C"
]

# Ordinal categorical features
ordinal_cols = ["C_H"]

# Binary categorical features
binary_cols = [
    "gender",
    "type_policy",
    "reimbursement",
    "new_business"
]

# Numerical features
categorical_cols = onehot_cols + ordinal_cols + binary_cols

numeric_cols = [
    col for col in X_train.columns
    if col not in categorical_cols
]

# Verify feature classification
all_features = numeric_cols + categorical_cols

assert len(all_features) == len(set(all_features))
assert set(all_features) == set(X_train.columns)
assert len(all_features) == 29

# Verify binary columns
for col in binary_cols:
    unique_values = X_train[col].dropna().unique()

    assert len(unique_values) == 2, (
        f"{col} has {len(unique_values)} categories: {unique_values}"
    )

# Summary
print("Numerical features:", len(numeric_cols))
print("One-Hot features:", len(onehot_cols))
print("Ordinal features:", len(ordinal_cols))
print("Binary features:", len(binary_cols))
print("Total predictors:", len(all_features))

# Inspect actual categories
for col in categorical_cols:
    print(f"\n{col}:")
    print(X_train[col].dropna().unique())
```

```python
# 1. Numerical preprocessing
numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])


# 2. Nominal categorical preprocessing
onehot_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(
        drop="first",
        handle_unknown="ignore",
        sparse_output=False
    ))
])


# 3. Ordinal categorical preprocessing
ordinal_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OrdinalEncoder(
        categories=[
            ["H1", "H2", "H3", "H4", "H5", "H6"]
        ],
        handle_unknown="use_encoded_value",
        unknown_value=-1
    ))
])


# 4. Binary categorical preprocessing
binary_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OrdinalEncoder(
        handle_unknown="use_encoded_value",
        unknown_value=-1
    ))
])

print("All four preprocessing pipelines defined successfully.")
```

```python
scaled_preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", numeric_pipeline, numeric_cols),

        ("onehot", onehot_pipeline, onehot_cols),

        ("ordinal", Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OrdinalEncoder(
                categories=[
                    ["H1", "H2", "H3", "H4", "H5", "H6"]
                ],
                handle_unknown="use_encoded_value",
                unknown_value=-1
            )),
            ("scaler", StandardScaler())
        ]), ordinal_cols),

        ("binary", binary_pipeline, binary_cols)
    ],
    remainder="drop",
    verbose_feature_names_out=True
)


# 2. Tree-based preprocessor
# For Decision Tree and Random Forest

tree_preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", SimpleImputer(strategy="median"), numeric_cols),

        ("onehot", onehot_pipeline, onehot_cols),

        ("ordinal", ordinal_pipeline, ordinal_cols),

        ("binary", binary_pipeline, binary_cols)
    ],
    remainder="drop",
    verbose_feature_names_out=True
)

print("Scaled preprocessor created successfully.")
print("Tree-based preprocessor created successfully.")
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

def regression_metrics(y_true, y_pred):
    return {
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
        "R²": r2_score(y_true, y_pred)
    }

results_lr = pd.DataFrame({
    "Training": regression_metrics(y_train, y_train_pred_lr),
    "Validation": regression_metrics(y_val, y_val_pred_lr)
}).T

display(results_lr.round(4))
```

```python
from sklearn.metrics import mean_absolute_percentage_error

# MAPE is undefined when actual premium is zero
train_mask = y_train.to_numpy() != 0
val_mask = y_val.to_numpy() != 0

# Training MAPE
train_mape = mean_absolute_percentage_error(
    y_train.to_numpy()[train_mask],
    y_train_pred_lr[train_mask]
) * 100

# Validation MAPE
val_mape = mean_absolute_percentage_error(
    y_val.to_numpy()[val_mask],
    y_val_pred_lr[val_mask]
) * 100

print(f"Training MAPE   : {train_mape:.2f}%")
print(f"Validation MAPE : {val_mape:.2f}%")
print(f"Excluded zero targets (train): {(~train_mask).sum()}")
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    mean_absolute_percentage_error,
    r2_score
)

import numpy as np
import pandas as pd

def evaluate_model(y_true, y_pred):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)

    nonzero = y_true != 0

    return {
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": np.sqrt(mean_squared_error(y_true, y_pred)),
        "R²": r2_score(y_true, y_pred),
        "MAPE (%)": (
            mean_absolute_percentage_error(
                y_true[nonzero],
                y_pred[nonzero]
            ) * 100
            if nonzero.any() else np.nan
        )
    }

results_lr_no_prev = pd.DataFrame({
    "Training": evaluate_model(y_train, y_train_pred_lr),
    "Validation": evaluate_model(y_val, y_val_pred_lr)
}).T

display(results_lr_no_prev.round(4))
```

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.base import clone
import pandas as pd

# Rebuild preprocessing without previous_premium
dt_preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", clone(numeric_pipeline), numeric_cols),
        ("onehot", clone(onehot_pipeline), onehot_cols),
        ("ordinal", clone(ordinal_pipeline), ordinal_cols),
        ("binary", clone(binary_pipeline), binary_cols)
    ],
    remainder="drop"
)

# Decision Tree does not require numerical scaling
dt_preprocessor.set_params(
    numeric__scaler="passthrough"
)

# Build model pipeline
dt_pipeline = Pipeline([
    ("preprocessor", dt_preprocessor),
    ("model", DecisionTreeRegressor(
        max_depth=8,
        min_samples_leaf=20,
        random_state=42
    ))
])

# Train
dt_pipeline.fit(X_train, y_train)

# Predictions
y_train_pred_dt = dt_pipeline.predict(X_train)
y_val_pred_dt = dt_pipeline.predict(X_val)

# Evaluation (uses previously defined evaluate_model)
results_dt = pd.DataFrame({
    "Training": evaluate_model(y_train, y_train_pred_dt),
    "Validation": evaluate_model(y_val, y_val_pred_dt)
}).T

display(results_dt.round(4))
```

```python
# ============================================================
# RANDOM FOREST REGRESSION — BASELINE
# ============================================================

from sklearn.ensemble import RandomForestRegressor
from sklearn.pipeline import Pipeline
from sklearn.base import clone
import pandas as pd

# Use the existing tree preprocessor
# (No scaling required for Random Forest)

rf_pipeline = Pipeline([
    ("preprocessor", clone(dt_preprocessor)),
    ("model", RandomForestRegressor(
        n_estimators=100,
        max_depth=12,
        min_samples_split=10,
        min_samples_leaf=5,
        max_features=0.8,
        n_jobs=-1,
        random_state=42
    ))
])

# Train Random Forest
rf_pipeline.fit(X_train, y_train)

# Predictions
y_train_pred_rf = rf_pipeline.predict(X_train)
y_val_pred_rf = rf_pipeline.predict(X_val)

# Evaluation
results_rf = pd.DataFrame({
    "Training": evaluate_model(y_train, y_train_pred_rf),
    "Validation": evaluate_model(y_val, y_val_pred_rf)
}).T

display(results_rf.round(4))
```

```python
import optuna
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import GroupKFold, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.base import clone

# ------------------------------------------------------------
# 1. Group-aware cross-validation
# ------------------------------------------------------------

groups_train = train_df.loc[X_train.index, "ID_policy"]

cv = GroupKFold(n_splits=3)

# ------------------------------------------------------------
# 2. Define optimization objective
# ------------------------------------------------------------

def objective(trial):

    params = {
        "n_estimators": trial.suggest_int(
            "n_estimators", 60, 140, step=20
        ),
        "max_depth": trial.suggest_int(
            "max_depth", 6, 22
        ),
        "min_samples_split": trial.suggest_int(
            "min_samples_split", 4, 30
        ),
        "min_samples_leaf": trial.suggest_int(
            "min_samples_leaf", 5, 40
        ),
        "max_features": trial.suggest_float(
            "max_features", 0.5, 1.0
        ),
    }

    pipeline = Pipeline([
        ("preprocessor", clone(dt_preprocessor)),
        ("model", RandomForestRegressor(
            **params,
            random_state=42,
            n_jobs=-1
        ))
    ])

    scores = cross_val_score(
        pipeline,
        X_train,
        y_train,
        groups=groups_train,
        cv=cv,
        scoring="neg_mean_absolute_error",
        n_jobs=1
    )

    return -scores.mean()

# ------------------------------------------------------------
# 3. Run Bayesian Optimization
# ------------------------------------------------------------

optuna.logging.set_verbosity(optuna.logging.WARNING)

sampler = optuna.samplers.TPESampler(seed=42)

study = optuna.create_study(
    direction="minimize",
    sampler=sampler
)

study.optimize(
    objective,
    n_trials=20,
    show_progress_bar=True
)

# ------------------------------------------------------------
# 4. Display best results
# ------------------------------------------------------------

print("Best Cross-Validation MAE:", round(study.best_value, 4))

print("\nBest Hyperparameters:")
for parameter, value in study.best_params.items():
    print(f"{parameter}: {value}")
    
```
